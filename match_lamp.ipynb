We want to match our [lamp_b_binned_sub_bias.ms.fits](lamp_b_binned_sub_bias.ms.fits) (coupled with the ashi list) with our image 
[image1_calibrated.ms.fits](image1_calibrated.ms.fits). \
We use the routine refspectra.

```python
import os
import shutil
from pyraf import iraf

# 1. Copy lamp file on this folder
shutil.copy("lamp/lamp_b_binned_sub_bias.ms.fits", ".")

# 2. Copy also lamp database folder (if exists)
if os.path.exists("lamp/database"):
    shutil.copytree("lamp/database", "database", dirs_exist_ok=True)

# 3. Execute refspectra directly on the local file
iraf.noao()
iraf.onedspec()

iraf.refspectra(
    input="image1_calibrated.ms.fits",
    references="lamp_b_binned_sub_bias.ms.fits",
    apertures="",
    ignoreaps="yes",
    sort="",
    group="",
    confirm="no",
)

#this modified the header
```

```python
import glob
import os
import shutil
from pyraf import iraf

iraf.flpr()

iraf.noao()
iraf.onedspec()

file_ms = "image1_dispcor.ms.fits"
cartella_out = "SN2024gy_spectra"
os.makedirs(cartella_out, exist_ok=True)

# Crea un file cursore temporaneo con il comando 'q' (quit) per chiudere splot subito dopo aver disegnato
cursor_file = "temp_cursor.txt"
with open(cursor_file, "w") as f:
    f.write("q\n")

for ap in range(1, 15):
    # Il parametro corretto in splot è graphics="psdump"
    iraf.splot(
        images=file_ms, line=ap, graphics="psdump", cursor=cursor_file
    )
    iraf.gflush()

    # Recupera l'ultimo file .ps generato in /tmp/ e lo sposta nella cartella SN2024gy_spectra
    files_ps = sorted(glob.glob("/tmp/irafdmp*.ps"), key=os.path.getmtime)
    if files_ps:
        destinazione = os.path.join(
            cartella_out, f"image1_dispcor_ap{ap:02d}.ps"
        )
        shutil.move(files_ps[-1], destinazione)
        print(f"Salvato: {destinazione}")

# Rimuove il file cursore temporaneo
os.remove(cursor_file)
```

```python
import glob
import os
import subprocess

cartella = "SN2024gy_spectra"
files_ps = sorted(glob.glob(os.path.join(cartella, "*.ps")))

for ps_path in files_ps:
    png_path = ps_path.replace(".ps", ".png")
    # Usa Ghostscript (standard su Linux/macOS) ruotando l'immagine in orizzontale
    subprocess.run(
        [
            "gs",
            "-dSAFER",
            "-dBATCH",
            "-dNOPAUSE",
            "-sDEVICE=png16m",
            "-r200",
            f"-sOutputFile={png_path}",
            ps_path,
        ],
        check=True,
    )
    print(f"Convertito: {png_path}")
```

```python
import glob
import os
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

cartella = "SN2024gy_spectra"
files_png = sorted(
    glob.glob(os.path.join(cartella, "image1_dispcor_ap*.png"))
)

n_files = len(files_png)
print(f"Trovati {n_files} file PNG")

# Larghezza grande (20 pollici) per occupare tutto lo schermo, altezza proporzionale (10 pollici per grafico)
fig, axes = plt.subplots(
    nrows=n_files, ncols=1, figsize=(20, 10 * n_files), gridspec_kw={"hspace": 0.08}
)

if n_files == 1:
    axes = [axes]

for idx, png_path in enumerate(files_png):
    ax = axes[idx]
    img = mpimg.imread(png_path)

    # Ruota di -90° se l'immagine è in verticale (altezza > larghezza)
    if img.shape[0] > img.shape[1]:
        img = img.swapaxes(0, 1)[:, ::-1, :]

    ax.imshow(img, aspect="auto")
    ax.axis("off")
    ax.set_title(
        os.path.basename(png_path),
        fontsize=14,
        fontweight="bold",
        loc="left",
        pad=4,
    )

plt.subplots_adjust(left=0.01, right=0.99, top=0.99, bottom=0.01)
plt.savefig(
    os.path.join(cartella, "tutte_14_aperture_verticale.png"),
    dpi=200,
    bbox_inches="tight",
)
plt.show()
```
