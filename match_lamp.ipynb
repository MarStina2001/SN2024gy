We want to match our [lamp_b_binned_sub_bias.ms.fits](lamp_b_binned_sub_bias.ms.fits) (coupled with the ashi list) with our image 
[image1_calibrated.ms.fits](image1_calibrated.ms.fits). \
We use the routine refspectra.

```python
import os
import shutil
from pyraf import iraf

# 1. Copy lamp file on this folder
shutil.copy("lamp/lamp_b_binned_sub_bias.ms.fits", ".")

# 2. Copy also lamp database folder (if exists)
if os.path.exists("lamp/database"):
    shutil.copytree("lamp/database", "database", dirs_exist_ok=True)

# 3. Execute refspectra directly on the local file
iraf.noao()
iraf.onedspec()

iraf.refspectra(
    input="image1_calibrated.ms.fits",
    references="lamp_b_binned_sub_bias.ms.fits",
    apertures="",
    ignoreaps="yes",
    sort="",
    group="",
    confirm="no",
)

#this modified the header
```

Trasferiamo pixels a lunghezze d'onda cojn dispcor [image1_calibrated.ms.fits](image1_calibrated.ms.fits) ---> [image1_dispcor.ms.fits](image1_dispcor.ms.fits):

```python
from astropy.io import fits
from pyraf import iraf

# Ora lancia specificamente il dispcor del pacchetto echelle
iraf.noao()
iraf.imred()
iraf.echelle()

iraf.echelle.dispcor(
    input="image1_calibrated.ms.fits",
    output="image1_dispcor.ms.fits",
)

#da errore perché non ho fittato le righe dopo averle identificate con ecidentify
```
