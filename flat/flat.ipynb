Before starting I selected all the flats with "Catherine" instrument (visible light). So the ccdtype filtering is not necessary.\
Let's calculate the average of flats with the routine flatcombine of Iraf.

```python
import glob
import os
from pyraf import iraf

# 0. If you Windows operative system, be sure that in your input file_list.txt there is _ and not :
# also replace the middle point "·" with "_" on the files automatically dowloaded by the ESO archive into C or D Windows hard disks.

# 1. Load the necessary IRAF packages
iraf.noao()
iraf.imred()
iraf.ccdred()

# 2. Make sure to adjust the pattern to match your actual flat file names
flat_pattern = "XSHOO*.fits"
flat_files = sorted(glob.glob(flat_pattern))

if not flat_files:
    raise FileNotFoundError(f"No files found with pattern: {flat_pattern}")

# If the data comes from MEF (multi-extension) instruments like X-Shooter,
# specify the scientific extension for each file (e.g., bias.fits[1])
# bias_files = [f"{f}[1]" for f in bias_files]

input_list = "flat_list_forWindows.txt"
with open(input_list, "w") as f:   #it's writing file names into the list automatically 
    for filename in flat_files:
        f.write(f"{filename}\n")

# 3. Name of the output Master Flat file
output_master_flat = "Flat.fits"

# Remove the previous output if it already exists
if os.path.exists(output_master_flat):
    os.remove(output_master_flat)

# 4. Run flatcombine using PyRAF
# Note: combine='median' or 'average' with reject='sigclip' or 'minmax'
iraf.flatcombine(    
    input=f"@{input_list}",
    output=output_master_flat,
    combine="median",       # 'median' or 'average'
    reject="sigclip",      # 'sigclip' or 'minmax'
    ccdtype="",            # ESSENTIAL: Prevents IRAF from filtering out ESO headers
    process="no",          # Do not run ccdproc
    subsets="no",          # Disable filter/subset grouping checks
    scale="median",        # Scale flat fields by their median level before combining
    statsec="",            # Full frame
)

print(f"Master Flat created successfully: {output_master_flat}")

# 5. Cleanup (optional)
#if os.path.exists(input_list):
#    os.remove(input_list)
```
